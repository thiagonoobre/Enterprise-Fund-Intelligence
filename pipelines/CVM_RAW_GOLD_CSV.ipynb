### CVM Fundos diarios

```python
# Mais performatica para pegar os metadados
show_partitions_df = spark.sql("SHOW PARTITIONS workspace.case_spark_cvm.silver_cvm_fundos_diario")

# maior partição da data_processamento
max_partition = show_partitions_df.agg(f.max("data_processamento")).collect()[0][0]
print("Max data_processamento partition:", max_partition)

# pegar o dataframe com maior partição
df_cvm_fundos_diario = spark.table("workspace.case_spark_cvm.silver_cvm_fundos_diario")\
                        .filter(f.col("data_processamento") == max_partition)
```

```python
df_cvm_fundos_diario.show()
```
